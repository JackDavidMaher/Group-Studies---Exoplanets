```python
import numpy as np
from scipy.interpolate import RegularGridInterpolator

# Load files
A_H_data = np.load("../Group 1 Full Loop Code/exoplanetAH_data.npy")
Tg_data = np.load("../Group 1 Full Loop Code/exoplanetTg_data.npy")

# Extract unique grid axes
Temps = np.unique(Tg_data[:,0])      # unique temperatures
gravities = np.unique(Tg_data[:,1])  # unique gravities

# print("Temps:", Temps.shape, "gravities:", gravities.shape)

# Create interpolator
interp = RegularGridInterpolator(
    (Temps, gravities),
    A_H_data,
    method='linear',
    bounds_error=False,
    fill_value=None
)

# Example usage
T_test, g_test = 500, 11
Ah_value = interp([[T_test, g_test]])
print("Interpolated AH:", Ah_value)
```
