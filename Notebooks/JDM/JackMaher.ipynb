```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Wedge

fig, ax = plt.subplots(figsize=(8, 6))

planet_radius = 0.7
outer_radius = 2
n_shells = 15

# Atmospheric shell boundaries from outer edge to planet surface.
radii = np.linspace(outer_radius, planet_radius, n_shells + 1)
colors = plt.cm.Blues(np.linspace(0.25, 0.9, n_shells))

# Draw semicircular atmosphere layers (cross-section).
for i in range(n_shells):
    shell = Wedge(
        center=(0, 0),
        r=radii[i],
        theta1=0,
        theta2=180,
        width=radii[i] - radii[i + 1],
        facecolor=colors[i],
        edgecolor=colors[i],
    )
    ax.add_patch(shell)

# Draw the planet beneath the atmosphere.
planet = Wedge(
    center=(0, 0),
    r=planet_radius,
    theta1=0,
    theta2=180,
    facecolor="#000000",
    edgecolor="#000000",
    linewidth=1.2,
    zorder=3,
 )
ax.add_patch(planet)

# Subtle guides to separate layers visually.
angles = np.linspace(0, np.pi, 300)
for r in radii[1:-1]:
    ax.plot(
        r * np.cos(angles),
        r * np.sin(angles),
        color="white",
        linewidth=0.7,
        alpha=0.35,
    )

# Red lines in the atmosphere that get dimmer lower in the diagram.
line_ys = np.linspace(planet_radius + 0.03, outer_radius - 0.03, 15)
x_edges = np.sqrt(outer_radius**2 - line_ys**2)

# Two extra line sets with vertical outer ends.
extension_length = 0.32
x_outer = np.max(x_edges) + extension_length

line_min = np.min(line_ys)
line_span = np.ptp(line_ys)

for y, x_edge in zip(line_ys, x_edges):
    t = (y - line_min) / line_span
    line_color = plt.cm.Reds(0.25 + 0.7 * t)
    line_alpha = 0.35 + 0.3 * t

    ax.hlines(
        y,
        -x_edge,
        x_edge,
        colors=[line_color],
        linewidth=1.0,
        alpha=1,
        zorder=4,
    )
    ax.hlines(
        y,
        -x_outer,
        -x_edge,
        colors="#FF0000",
        linewidth=1.0,
        alpha=1,
        zorder=4,
    )
    ax.hlines(
        y,
        x_edge,
        x_outer,
        colors=[line_color],
        linewidth=1.0,
        alpha=line_alpha,
        zorder=4,
    )
    # Small right-pointing arrows on both left and right side extensions.
    arrow_dx = 0.08
    arrow_inset = 0.04

    left_start = -x_outer + arrow_inset
    if left_start + arrow_dx < -x_edge:
        ax.annotate(
            "",
            xy=(left_start + arrow_dx, y),
            xytext=(left_start, y),
            arrowprops=dict(
                arrowstyle="->",
                color="#FF0000",
                lw=0.9,
                mutation_scale=8,
                shrinkA=0,
                shrinkB=0,
            ),
            zorder=5,
        )

x_extent = x_outer + 0.1
ax.set_xlim(-x_extent, x_extent)
ax.set_ylim(0.4, outer_radius * 1.15)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title("2D Atmospheric Cross-Section")

ax.text(
    0.02,
    0.94,
    "higher rays are\nnot absorbed",
    transform=ax.transAxes,
    ha="center",
    va="center",
    fontsize=9,
    color="#000000",
    bbox=dict(
        boxstyle="round,pad=0.25",
        facecolor="#ffffff",
        edgecolor="#000000",
        alpha=0.9,
    ),
)

ax.text(
    0.02,
    0.1,
    "lower rays are\nblocked by clouds",
    transform=ax.transAxes,
    ha="center",
    va="center",
    fontsize=9,
    color="#000000",
    bbox=dict(
        boxstyle="round,pad=0.25",
        facecolor="#FFFFFF",
        edgecolor="#000000",
        alpha=0.9,
    ),
)

plt.tight_layout()
plt.savefig("atmospheric_cross_section.png", dpi=300)
plt.show()
```
